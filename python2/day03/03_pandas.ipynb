```python
# pandas 예제
import pandas as pd
```

```python
# 리스트를 이용 => Series
list = [100, 20, 30, 40]
# Series: 인덱스가 있는 1차원 배열
scores = pd.Series(list, index=["국어", "영어", "수학", "과학"])

print(scores)
# DataFrame : 행과 열로 이루어진 표 형태의 데이터구조
# 딕셔너리를 이용 => DataFrame
scores_dic = {
	'과목' : ["국어", "영어", "수학", "과학"],
	'성적' : [100, 90, 80, 70],
	'출석' : [100, 100, 0, 80],
	'수행' : [90, 100, 65, 30]
}
df = pd.DataFrame(scores_dic)
print(df)
```

```python
# 데이터 탐색 예제
print(f"상위 2행 : \n{df.head(2)}")
print("-"*20)
print(f"상위 2행 : \n{df.tail(2)}")
print("-"*20)
print(f"info : \n{df.info()}")
print("-"*20)
print(f"describe : \n{df.describe()}")
print("-"*20)
print(f"shape : \n{df.shape}")
print("-"*20)
```

```python
def bar():
	print("-"*30)
```

```python
# 데이터 선택 및 필터링
print("데이터 프레임")
print("-"*20)
print(df)
bar()

# 과목들
print(df['과목'])
bar()

# 국어 정보 선택
print(f"국어 정보 : \n{df.loc[0]}")
bar()

# 과목의 수행평가 점수 조회
print("과목과 수행 평가 조회")
# 조회하려는 열이 여러개인 경우 열 목록을 리스트로 전달
print(df[['과목', '수행']])
bar()

# 국어, 수학 정보 조회
print("국어, 영어, 수학 정보 조회")
print(df.iloc[[0,3]])
bar()

print(f"국어 성적 : {df.loc[0, '성적']}")

# 성적이 80이상인 데이터를 조회
print(df[df['성적'] >= 80])
print(df.loc[df['성적'] >= 80, '과목'])
print(df.loc[df['성적'] >= 80, '과목'])
print(df.query('성적 >= 80') ['과목'])
```

```python
# 데이터 결측치 처리
df2 = pd.DataFrame({
	'이름' : ['홍길동', '이순신', '고길동'],
	'나이' : ['21', 'None', '40'],
	'주소' : ['홍길동', '이순신', '고길동'],
	'가족' : ['홍길동', '이순신', '고길동']
})
print(df2)
```

```python
# 결측지 개수를 조회
print(df2.isnull().sum())
```

```python
# 결측치가 있는 행 삭제
print(df2.dropna())
```

```python
# 결측치를 특정값으로 채움
print(df2.fillna(0))
```

```python
# pandas 메서드는 원본을 건드리지 않음
print(df2)
# 원본은 = 를 통해 덮어써야함
```

```python
# 데이터 가공(수정, 삭제, 추가)
print(df)
```

```python
# 시험 열 추가
df['시험'] = df['성적']
print(df)
```

```python
# 성적 열 삭제
df.drop('성적', axis=1, inplace=True)
print(df)
```

```python
# 총점 열 추가
df['총점'] = df['출석'] + df['수행'] + df['시험']
print(df)
```

```python
# 비고 열 추가 및 값 추가
df['비고'] = '없음'
print(df)
```

```python
# 0번지 행 삭제
print(df.drop(0)) # axis 기본값 0 
```

```python
# 과목명이 과학인 행 삭제
index = df[df['과목'] == '과학'].index
print(df.drop(index))
```

```python
# 시험 성적 순으로 정렬
print(df.sort_values(by='시험'))
```

```python
# 시험 성적 순으로 정렬 : 내림차순
print((df.sort_values ascending=False))
```

```python
# 데이터 결합 및 그룹 예제
df3 = pd.DataFrame({
	"이름" : ["홍길동", "고길동", "둘리"],
	"국어성적" : ["홍길동", "고길동", "둘리"]
})

df4 = pd.DataFrame({
	"이름" : ["홍길동", "고길동", "둘리"],
	"영어성적" : [80, 70, 90]
})
```

```python
# 국어 성적과 영어 성적 합치기
total = pd.merge(df3, df4, on="이름")
print(total)
```
