```python
import pandas as pd
# 다음과 같이 리스트가 있을 때 pandas를 이용하여 표를 반환하세요.

names = ["홍길동", "둘리", "고길동"]
kor_scores = [100, 90, 100]
eng_scores = [90, 10, 0]

# 이름 국어 영어
# 홍길동 100 90
# 둘리 90 10
# 고길동 100 0

df = pd.DataFrame({
	'이름' : names,
	'국어' : kor_scores,
	'영어' : eng_scores
})

print(df)
```

```python
# 국어 성적 평균을 조회
kor_avg = df['국어'].mean()
print(f"국어 성적 평균 : {kor_avg}")
```

```python
# 평균 열을 추가

# df = pd.DataFrame({
# 	'이름' : names,
# 	'국어' : kor_scores,
# 	'영어' : eng_scores,
# 	'평균' : kor_scores + eng_scores / 2
# })

df['평균'] = (df['국어'] + df['영어']) / 2
print(df)

# 평균 열의 값을 국어와 영어의 평균으로 계산해서 넣어줌
```
