```python
# 카페 매출 정보를 가져오는 함수
import requests
import pandas as pd

def get_cafe_revenue():
	url = 'https://my-json-server.typicode.com/st8324/repo/cafe'

	try:
		response = requests.get(url)
		data = response.json()
		response.raise_for_status()
		print(data)
		list = []

		for item in data:
			dic = {
				'주문시간' : item['주문시간'],
				'메뉴' : item['메뉴'],
				'카테고리' : item['카테고리'],
				'수량' : item['수량'],
				'가격' : item['가격'],
				'결제수단' : item['결제수단']
			}
		
			list.append(dic)
			df = pd.DataFrame(list)
			print(df)

		return df

	except Exception as e:
		print(f"에러 발생 : {e}")

df = get_cafe_revenue()
```

```python
# 중복 및 결측치 처리하는 함수
df = df.drop_duplicates(subset=['주문시간'], keep='first')
```

```python
# 확인
print(df)
```

```python
# 데이터를 활용한 예제
# 매출을 조회하기 위해 '결제액'을 추가
# 결제액은 가격 * 수량
df['가격'] = pd.to_numeric(df['가격'])
df['수량'] = pd.to_numeric(df['수량'])

df['결제액'] = df['가격'] * df['수량']
print(df)
```

```python
# 전체 매출액을 조회
entire_revenue =  df['가격'] * df['수량']
print(entire_revenue)
```

```python
# 메뉴별 매출액을 조회(메뉴, 수량, 결제액)
# groupby
menu_df = df.groupby('메뉴')[['수량', '결제액']].sum()
print(menu_df)
```

```python
# 카테고리별 수량과 결재액을 조회
category_df = df.groupby('카테고리')[['수량', '결제액']].sum()
category_df = category_df.sort_values(by='결제액', ascending=False)

# 카테고리별 매출 비율을 추가
# 전체 매출액 : revenue
category_df['매출비율'] = category_df['결제액'] / entire_revenue * 100
category_df['매출비율'] = category_df['매출비율'].round(1)

# 매출이 가장 높은 카테고리를 조회
top_category = category_df['결제액'].max()
print(category_df.loc[category_df['결제액'] == top_category])
```

```python
# 문자열 시간을 시간 포맷으로 변환
df['주문시간'] = pd.to_datetime(df['주문시간'])
# 시간대를 추가
df['시간'] = df['주문시간'].dt.hour
print(df)

print("ㅡㅡㅡ" * 15)

# 시간대별 주문 건수를 조회
hour_df = df['시간'].value_counts()
print(hour_df)
```

```python
# 시간대별로 매출액을 조회 : 오전(12시 이전), 오후(17시 이전), 저녁(17시 이후)
# 시간을 시간대로 변환하는 함수
def get_time_slot(hour):
	if hour < 12:
		return "오전"
	elif hour < 17:
		return "오후"
	else:
		return "저녁"

# 시간을 시간대로 변경한 열 추가
df['시간대'] = df['시간'].apply(get_time_slot)

# 시간대별 결제액을 조회
time_df = df.groupby('시간대')['결제액'].sum()
print(time_df)
```

```python
# 요일별 결제액 조회
df['요일'] = df['주문시간'].dt.day_of_week
print(df)
print("ㅡ" * 50)
df['요일'] = df['주문시간'].dt.day_name()

# 요일을 한글로
dayweek_map = {
	'Monday' : '월요일',
	'Tuesday' : '화요일',
	'Wednesday' : '수요일',
	'Thursday' : '목요일',
	'Friday' : '금요일',
	'Saturday' : '토요일',
	'Sunday' : '일요일',
}
df['요일'] = df['요일'].map(dayweek_map)
print(df)
```

```python
# 전체를 카페_매출.csv로 저장
entire_revenue.to_csv('카페_매출.csv', encoding='utf-8-sig', index=False)
# 메뉴별 매출을 카페_메뉴_매출.csv로 저장
menu_df.to_csv('카페_메뉴_매출.csv', encoding='utf-8-sig', index=False)
# 카테고리별 매출을 카페_카테고리_매출.csv로 저장
category_df.to_csv('카페_카테고리_매출.csv', encoding='utf-8-sig', index=False)
# 시간대별 매출을 카페_시간_매출.csv로 저장
time_df.to_csv('카페_시간_매출.csv', encoding='utf-8-sig', index=False)
# 요일별 매출을 카페_요일_매출.csv로 저장
df['요일'].to_csv('카페_요일_매출.csv', encoding='utf-8-sig', index=False)
```
