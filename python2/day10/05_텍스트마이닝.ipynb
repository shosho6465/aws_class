### 단어 가방 (Bag of Words : BoW)
* 문장을 단어와 출현 회수로 이루어진 고유한 맵으로 변환하여 백터화 하는 과정

```python
# 문장들 생성
corpos = ['강사님은 파이썬을 좋아합니다',
					'강사님은 자바도 좋아합니다.',
					'파이썬은 어렵습니다.',
					'학생들도 파이썬을 좋아합니다']
```

```python
from sklearn.feature_extraction.text import CountVectorizer

vectorizer = CountVectorizer()
# BoW 변환
bow_vector = vectorizer.fit_transform(corpos)
```

```python
vectorizer.vocabulary_ # 추출한 단어 목록들
```

```python
bow_vector.toarray()
# 강사님은 어렵습니다 자바도 좋아합니다 파이썬은 파이썬을 학생들도
```

### TF-IDF(Term Frequency-Inverse Document Frequency)
* 반복되서 나오는 단어는 중요하지만, 모든 문서에서 나오는 반복된 단어는 중요도가 낮음
* TF : 단어가 해당 문서에 얼마나 등장하는가? => 많을수록 좋음
* IDE : 해당 단어가 얼마나 많은 문서에 공통으로 등장하는가? => 많을수록 감점


```python
# TD_IDF 예제
```

```python
# 데이터 생성
corpus = ['강사님은 파이썬을 좋아합니다',
					'강사님은 자바도 좋아합니다',
					'파이썬은 어렵습니다',
					'학생들도 파이썬을 좋아합니다']
```

```python
okt = Okt()

# 주어진 문자열들을 명사로만된 문자열들로 바꾸는 과정
# 형태소 분석 및 명사 추출
clean_corpus = []
for text in corpus:
	# 명사들만 추출
	nouns = okt.nouns(text)
	# 명사들을 하나의 문자열로 변환
	clean_text = " ".join(nouns)
	# 결과 리스트에 추가
	clean_corpus.append(" ".join(clean_text))
	
print(clean_corpus)
```

```python
# 명사, 동사, 형용사를 골라내는 작업
clean_corpus2 = []
text = '강사님은 파이썬을 좋아합니다'

# stem=True : 동사를 동사원형을 바꿈
result = okt.pos(text, stem=True)
words = []
print(result)

for word, pos in result:
	print(f'{word}({pos})')

	# 단어의 품사가 명사, 동사, 형용사이면 추출하여 리스트에 추가
	if pos in ['Noun', "Verb", 'Adjective']:
		words.append(word)
# 리스트에 있는 단어들을 문자열로 변환하여 명사, 동사, 형용사만 있는 문자열로 변환
new_text = " ".join(words)

new_text
```
