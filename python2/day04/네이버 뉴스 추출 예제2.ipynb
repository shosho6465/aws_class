```python
# import np # 네이버 뉴스를 검색해서 제목과 링크를 가져오는 모듈
import time
# csv 파일에 있는 뉴스 링크로 가서 기사 내용을 가져와서 naver_nl_ai_1_detail.csv를 생성하는 예제
# naver_nl_ai_페이지_detail.csv 열 이름 => title, href, detail

import requests
from bs4 import BeautifulSoup as bf
import pandas as pd

# keyword = 'ai'
url = f'https://n.news.naver.com/mnews/article/011/0004598795?sid=105'

headers = {
	"User-Agent" : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/145.0.0.0 Safari/537.36',
	"Referer" : 'https://search.naver.com/search.naver'
}

try:
	response = requests.get(url)
	response.raise_for_status()
	print("연결 성공!")

	soup = bf(response.text, 'lxml')
	content_link = soup.select_one('.go_trans._article_content')

	# title_link = soup.select_one()
	content = content_link.text

	df = pd.DataFrame({
		'content' : [content]
	})
	print(df)

	df.to_csv(f'naver_nl_content.csv', index=False, encoding='utf-8-sig')


except Exception as e:
	print(f"예외 발생 : {e}")
```
