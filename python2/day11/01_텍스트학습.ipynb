```python
# 데이터 생성
import pandas as pd
csv = pd.read_csv('sample.csv')
csv.head(5)
print(csv)
```

```python
#  비어있는 행 삭제
csv.dropna()
```

```python
# 형태소 분석. 명사, 동사, 형용사를 추출
from konlpy.tag import Okt
okt = Okt()

sentences = csv['sentence'].tolist()
print(sentences)
cleaned_sentences = []

for sentence in sentences:
	result = okt.pos(sentence, stem=True)
	words = [word for word, pos in result if pos in ['Noun', 'Verb', 'Adjective']]

	# 리스트에 있는 단어들을 문자열로 변환 => 명사, 동사, 형용사만 있는 문자로 변환
	new_text = " ".join(words)
	cleaned_sentences.append(new_text)

print(cleaned_sentences) 


# 결과물 id, sentence, label, cleaned_sentence
csv['cleaned_sentence'] = cleaned_sentences
csv
```

```python
# 텍스트 전처리 함수
# 문장에서 명사, 동사, 형용사면 원형을 추출해서 문장을 만들어 반환
def text_preprocessing(text):
	result = okt.pos(text, stem=True)
	words = [word for word, pos in result if pos in ['Noun', 'Verb', 'Adjective']]
	return " ".join(words)
```

```python
text_preprocessing('오늘 날씨가 정말 화창해서 기분이 최고예요')
# 오늘 날씨 정맣 화창하다 기분 최고

csv['cleaned_sentence2'] = csv['sentence'].apply(text_preprocessing)
csv[['cleaned_sentence', 'cleaned_sentence2']]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(csv['cleaned_sentence'])
y = csv['label']

# 원래는 훈련, 테스트를 분리하여 학습 후의 정확도를 확인
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

```python
# 학습
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python
# 테스트
text = '자격증 시험에 합격했습니다'
vector_text = vectorizer.transform([text])
res = model.predict(vector_text)
print(f'{text} : {'긍정' if res[0] else '부정'}')

# 시험 떨어지다라는 데이터를 학습하지 못해서 결과가 정확하지 않음
text = '날씨가 화창한데 자격증 시험에 떨어졌습니다'
vector_text = vectorizer.transform([text])
res = model.predict(vector_text)
print(f'{text} : {'긍정' if res[0] else '부정'}')
```
