```python
# 딕셔너리 선언
dic = {
    'name' : "홍길동",
    'age' : 21,
    'addr' : "서울시"
}
print(dic)
# 이름 확인
print(f"이름 : {dic["name"]}") #키가 없는 경우 에러 발생
print(f"이름 : {dic.get("name", "익명")}") # 키가 없어도 기본값이 실행됨 => 에러 발생 없이 안전하게 가져옴
# 새로운 키 추가
dic["job"] = "강사"
print(dic)
```

```python
# set : {}를 선언
set_a = {"바나나", "사과", "딸기", "포도"}
print(set_a)

set_b = { "바나나", "귤", "오렌지", "딸기"}
print(set_b)
```

```python
#합집합 : |, union()
# "바나나", "딸기", "포도", "사과", "귤", "바나나", "오렌지", "딸기"
# => 
print("A와 B의 합집합")
print(set_a | set_b) 
print(set_a.union(set_b))
```

```python
# 교집합 : &, intersection()
print("A와 B의 교집합 : 두 집합에 공통으로 있는 값들")
print(set_a & set_b)
print(set_a.intersection(set_b))
```

```python
#차집합 : -, diffrence()
print("A - B: 집합 B에 없고 A에만 있는 값들")
print(set_a - set_b)
print(set_a.difference(set_b))
```

```python
# 대칭 차집합 : ^
# 겹치지 않은 모든 원소들 모음
# (A - B) | (B - A)
print(set_a ^ set_b) #(A | B) - (A & B)
```

```python
# 집합 예제(조건, 이름은 중복 x)
# 수학을 듣는 학생 목록
course_math = {"홍길동", "임꺽정", "둘리", "고길동", "하니"}
# 국어를 듣는 학생 목록
course_kor = {"이순신", "이도", "김삿갓", "둘리", "홍길동"}

# 수학만 듣는 학생 목록을 조회
print(f"수학만 듣는 학생 목록: {course_math - course_kor}")

# 수학, 국어 모두 듣는 학생 목록을 조회
print(f"수학, 국어 모두 듣는 학생 목록 : {course_math & course_kor}")

# 수학 또는 국어를 듣는 학생 목록을 조회
print(f"수학 또는 국어를 듣는 학생 목록을 조회 : {course_math | course_kor}")

#수학이나 국어를 하나만 듣는 학생을 조회
print(f"수학이나 국어를 하나만 듣는 학생을 조회: {course_math ^ course_kor}") 
```
