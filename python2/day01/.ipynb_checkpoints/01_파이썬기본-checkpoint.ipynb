```python
# 두 변수를 선언
a, b = 10, 20
print(a,b)
```

```python
# 문자열 인덱싱
# java는 인덱스에 음수를 넣지 못하지만 파이썬은 가능하다.
str = "Hello python."
# 0: 첫번째 글자, -1: 뒤에서 첫번째 글자, -2: 뒤에서 두번째 글자
print(str[0], str[-1], str[-2])
```

```python
#문자열 슬라이싱 => 부분 문자열
#기존 문자열 전체(글자가 총 13자)
print(str[0:13])
print(str[0:])
#마지막 글자를 제외(글자가 총 13자)
print(str[0:12])

# python만 출력
print(str[6:])
print(str[6:12])
```

```python
#문자열 포맷팅
age, name = 20, "홍길동"
msg = f"{name}의 나이는 {age}입니다."
print(msg)
```
